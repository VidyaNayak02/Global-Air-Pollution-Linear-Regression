```python
numeric_columns = [
    "AQI Value",
    "CO AQI Value",
    "Ozone AQI Value",
    "NO2 AQI Value",
    "PM2.5 AQI Value"
]

for column in numeric_columns:
    plt.figure(figsize=(8, 5))
    plt.hist(df[column], bins=30, edgecolor="black")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title(f"Distribution of {column}")
    plt.show()
```

```python
df.duplicated().sum()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["CO AQI Value"], df["AQI Value"])
plt.xlabel("CO AQI Value")
plt.ylabel("AQI Value")
plt.title("CO AQI Value vs AQI Value")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Ozone AQI Value"], df["AQI Value"], alpha=0.5)
plt.xlabel("Ozone AQI Value")
plt.ylabel("AQI Value")
plt.title("Ozone AQI Value vs AQI Value")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["NO2 AQI Value"], df["AQI Value"], alpha=0.5)
plt.xlabel("NO2 AQI Value")
plt.ylabel("AQI Value")
plt.title("NO2 AQI Value vs AQI Value")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["PM2.5 AQI Value"], df["AQI Value"], alpha=0.5)
plt.xlabel("PM2.5 AQI Value")
plt.ylabel("AQI Value")
plt.title("PM2.5 AQI Value vs AQI Value")
plt.show()
```

```python
correlation = df[
    ["AQI Value", "CO AQI Value", "Ozone AQI Value",
     "NO2 AQI Value", "PM2.5 AQI Value"]
].corr()["AQI Value"]

print(correlation)
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(
    df[["AQI Value", "CO AQI Value", "Ozone AQI Value",
        "NO2 AQI Value", "PM2.5 AQI Value"]].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

```python
plt.figure(figsize=(6, 8))
plt.boxplot(df["PM2.5 AQI Value"])
plt.ylabel("PM2.5 AQI Value")
plt.title("Boxplot of PM2.5 AQI Value")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df["PM2.5 AQI Value"], df["AQI Value"])
plt.xlabel("PM2.5 AQI Value")
plt.ylabel("AQI Value")
plt.title("PM2.5 AQI Value vs AQI Value")
plt.show()
```
